```python
def all_tracks(f):
    a_id = f[0].index('artists')
    n_id = f[0].index('track_name')
    pop_id = f[0].index('popularity')
    every = {}
    
    for i in range(1, len(f)):
        art = f[i][a_id]
        name = f[i][n_id]
        pop = f[i][pop_id]

        if art in every:
            every[art][1].add(name)
            num += 1
            pop_sum += int(pop)
        else:
            num = 1
            pop_sum = 0
            every[art] = [num, {name}, pop_sum]

    return every
```

```python
def most_popular(f):
    a_id = f[0].index('artists')
    n_id = f[0].index('track_name')
    every = {}
    
    for i in range(1, len(f)):
        art = f[i][a_id]
        name = f[i][n_id]

        if art in every:
            every[art].add(name)
        else:
            every[art] = {name}

    best_list = max(every, key=every.get)

    return best_list


most_common(data)
```
